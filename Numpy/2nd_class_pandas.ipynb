```python
left = pd.DataFrame({
    'ID' : [1,2,3],
    'Name' : ['Pandy', 'Fuzzy', 'Bamboo']
})

right = pd.DataFrame({
    'ID' : [1,2,4],
    'Hobby' : ['Climbing', 'Sleeping', 'Eating']
})

merged_df = pd.merge(left, right, on='ID', how='right')
print(merged_df)
```

```python
concatenated_df = pd.concat([left, right], ignore_index=True)
```
