## Sorting Algorithms

In Computer Science, a sorting algorithm is an algorithm that puts elements of a list an order. The most frequently used orders are numerical order and lexicographical order, and either ascending or descending.

```python
# Bubble Sort

def sortArray(nums):
        n = len(nums)

        # Bubble Sort
        for i in range(n):
            isSwap = False
            for j in range(n-i-1):
                if nums[j] > nums[j+1]:
                    # swap
                    temp = nums[j]
                    nums[j] = nums[j+1]
                    nums[j+1] = temp
                    isSwap = True

            if not isSwap:
                break

        return nums

print(sortArray([3, 2, 4, 6, 7, 1, 8, 11, 15]))
```

```python
# Insertion Sort 

def sortArray(nums):
    n = len(nums)

    for i in range(1,n):
        key = nums[i]
        j = i-1
        while j>=0 and nums[j] > key:
            nums[j+1] = nums[j]
            j-=1
        nums[j+1] = key
    return nums
```
