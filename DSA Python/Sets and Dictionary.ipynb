## Sets

```python
set1 = {2, 4, "hello", 2, "tuhin", "hello", 2, 4}

print(set1)
print(type(set1))
print(len(set1))
```

```python
list1 = [2, 3, 4, 2, 1, 23, 41, 23, 6, 4, 41, 85, 2, 7, 9]

# count number of unique elements in this list
set1 = set(list1)
print(len(set1))
print(set1)
```

```python
set1 = {"Hello", 4, 5, "Tuhin"}
set1.add(100)

print(set1)
```

```python
set1 = {"hello", 2, 4, "Tuhin"}

if "hello" in set1:
    print(True)
else:
    print(False)
```

```python
set1 = {1, 2, 3}
set2 = set1.copy()

set2.add(4)

print(set1)
print(set2)
```

```python
set1 = {1, 8, 5, 2, 4}
set2 = {8, 2, 9, 6}

# union
print(set1 | set2)
print(set1.union(set2))

# intersection
print(set1 & set2)
print(set1.intersection(set2))

# difference
print(set1-set2)
print(set1.difference(set2))

# symmetric difference
print(set1^set2)
print(set1.symmetric_difference(set2))
```

## Dictionary

```python
x = {}
print(type(x))
```

```python
dict1 = {
    1: "tuhin",
    "hello" : 100,
    10.21 : "hi",
    "key" : True
}

print(dict1)
print(type(dict1))
print(len(dict1))
```

```python
dict1 = {
    1: "tuhin",
    "hello" : 100,
    10.21 : "hi",
    "key" : True
}
print(dict1["key"])
print(dict1)
```

```python
dict1 = {
    1: "tuhin",
    2: "tasin",
    3: "ayat",
    4: "rayhan"
}

dict1[5] = "sifat"

print(dict1)
```

```python
dict1 = {
    1: "tuhin",
    2: "tasin",
    3: "ayat",
    4: "rayhan"
}

dict1.pop(2)
del dict1[1]


print(dict1)
```

```python
dict1 = {
    1: "tuhin",
    2: "tasin",
    3: "ayat",
    4: "rayhan"
}

for key, value in dict1.items():
    print(key, value)
```

```python
list1 = ["tuhin", "tasin", "ayat", "sifat", "jomur", "tasin", "ayat", "ayat", "jomur"]

freq = {}

for name in list1:
    if name not in freq:
        freq[name] = 1
    else:
        freq[name] += 1

print(freq)
```

```python
s = "physics wallah skilss"

freq = {}

for name in s:
    if name not in freq:
        freq[name] = 1
    else:
        freq[name] += 1

print(freq)
```
