```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Create nodes
a = Node(5)
b = Node(6)
c = Node(3)

# Link nodes
a.next = b
b.next = c

head = a   # Head points to the first node


# Traverse and print the linked list
current = head
while current:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```

```python
# Traverse linked list

def printLinkedList(head):
    curr = head

    while curr != None:
        print(curr.data, end=" ")
        curr = curr.next
printLinkedList(head)
```

```python
# Linked List Insertion at the beginning

newNode = Node(4)
newNode.next = head
head = newNode

current = head
while current:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```

```python
# Linked List Insertion at the end

newNode = Node(1)

curr = head
while curr.next != None:
    curr = curr.next

curr.next = newNode

printLinkedList(head)
```

```python
# insertion at the k_th index

k = 2
newNode = Node(9)

curr = head
for i in range(k-1):
    curr = curr.next

newNode.next = curr.next
curr.next = newNode

printLinkedList(head)
```

```python
# deletion the first Node

head = head.next
printLinkedList(head)
```

```python
# deletion the end Node

curr = head
while curr.next.next != None:
    curr = curr.next

curr.next = None
printLinkedList(head)
```

```python
 # delete at the k_th index

k = 2

curr = head
for i in range(k-1):
    curr = curr.next

curr.next = curr.next.next
printLinkedList(head)
```

## Doubly Linked List

```python
class DoublyNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

# Create nodes
a = DoublyNode(5)
b = DoublyNode(6)
c = DoublyNode(3)

# Link nodes
a.next = b
b.prev = a
b.next = c
c.prev = b

# preverse nodes


head = a   # Head points to the first node


# Traverse and print the linked list
current = head
while current:
    print(current.data, end=" <-> ")
    current = current.next
print("None")
```
