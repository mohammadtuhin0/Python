```python
# Data filtering

movies = df[df["type"] == "MOVIE"]
```

```python
movies.head()
```

```python
movies = movies.explode('production_countries')
```

```python
country_count = movies['production_countries'].value_counts()

print(country_count)
```

```python
top_country = movies['production_countries'].value_counts().idxmax()

print(top_country)
```

```python
# Grouping

df.groupby('release_year').size()
```

```python
# Country অনুযায়ী movie count

df.groupby('production_countries').size()
```

```python
# Genre অনুযায়ী average IMDb score

df.groupby('genres')['imdb_score'].mean()
```

```python
# Year অনুযায়ী average runtime

df.groupby('release_year')['runtime'].mean()
```
