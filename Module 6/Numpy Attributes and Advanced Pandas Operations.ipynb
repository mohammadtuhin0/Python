```python
import pandas as pd
import numpy as np
```

```python
data = pd.DataFrame({
    'Name' : ['Alice', 'Bob', np.nan, 'Daisy', 'Ethan'],
    'Age' : [25, np.nan, 29, 22, np.nan],
    'Salary' : [50000, 60000, np.nan, 58000, 62000]
})
```

```python
data
```

```python
# Null value Delete/Dropped

data_dropped = data.dropna()
```

```python
data_dropped
```

```python
data
```

```python
# Data Filled

data_filled = data.fillna({
    'Name' : 'Unknown',
    'Age' : data['Age'].mean(),
    'Salary' : 0
})
```

```python
data_filled
```

```python
# Vertical Stack (vstack)
# Horizontal Stack (hstack)
# Depth Stack (dstack)
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
```

```python
v_stacked = np.vstack([a, b])
print(v_stacked)
```

```python
a = np.array([[1, 2, 5], [3, 4, 9]])
b = np.array([[5, 6], [7, 8]])
```

```python
h_stacked = np.hstack([a, b])
print(h_stacked)
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
```

```python
d_stacked = np.dstack([a, b])
print(d_stacked)
```
